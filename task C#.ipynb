```
using System;

class Program
{
    static int Main(string[] args)
    {
        const int MaxMark = 100;
        
        Console.Write("Enter student age: ");
        string ageText = Console.ReadLine();

        Console.Write("Enter student mark: ");
        string markText = Console.ReadLine();

        int.TryParse(ageText, out int age);
        int.TryParse(markText, out int mark);

        if (mark < 0 || mark > MaxMark)
        {
            Console.WriteLine("Invalid mark! It must be between 0 and 100.");
            return 1;
        }

        double percentage = (double)mark / MaxMark * 100;
        Console.WriteLine($"Percentage: {percentage}%");
        Console.WriteLine($"Marks needed to reach {MaxMark}: {MaxMark - mark}");

        return 0;
    }
}

    
```

```
using System;

class Program
{
    static int Main(string[] args)
    {
        Console.Write("Enter student mark: ");
        string markText = Console.ReadLine();
        int.TryParse(markText, out int mark);

        string grade;

        if (mark >= 90)
        {
            grade = "Excellent";
        }
        else if (mark >= 80)
        {
            grade = "Very Good";
        }
        else if (mark >= 65)
        {
            grade = "Good";
        }
        else if (mark >= 50)
        {
            grade = "Pass";
        }
        else
        {
            grade = "Fail";
        }

        Console.WriteLine($"Grade: {grade}");

        if (mark >= 50)
        {
            Console.WriteLine("The student passed.");
        }
        else
        {
            Console.WriteLine("The student failed.");
        }

        return 0;
    }
}

```

```
using System;

class Program
{
    static int Main(string[] args)
    {
        Console.WriteLine("Select a department:");
        Console.WriteLine("1 - Computer Science");
        Console.WriteLine("2 - Information Systems");
        Console.WriteLine("3 - Artificial Intelligence");
        Console.Write("Your choice: ");

        string choiceText = Console.ReadLine();
        int.TryParse(choiceText, out int choice);

        switch (choice)
        {
            case 1:
                Console.WriteLine("Computer Science");
                break;
            case 2:
                Console.WriteLine("Information Systems");
                break;
            case 3:
                Console.WriteLine("Artificial Intelligence");
                break;
            default:
                Console.WriteLine("Invalid Department");
                break;
        }

        return 0;
    }
}

```

```
using System;

class Program
{
    static int Main(string[] args)
    {
        for (int i = 1; i <= 10; i++)
        {
            Console.WriteLine(i);
        }

        
        string correctPassword = "1234";
        int attempts = 0;

        while (attempts < 3)
        {
            Console.Write("Enter password: ");
            string password = Console.ReadLine();

            if (password == correctPassword)
            {
                Console.WriteLine("Access granted.");
                break;
            }

            attempts++;
            Console.WriteLine("Wrong password.");
        }

        if (attempts == 3)
        {
            Console.WriteLine("Too many attempts. Access denied.");
        }

        
        for (int i = 1; i <= 20; i++)
        {
            if (i % 2 != 0)
            {
                continue;
            }
            Console.WriteLine(i);
        }

        
        int choice;
        do
        {
            Console.WriteLine("1 - Say Hello");
            Console.WriteLine("2 - Show Date");
            Console.WriteLine("3 - Exit");
            Console.Write("Your choice: ");

            int.TryParse(Console.ReadLine(), out choice);

            switch (choice)
            {
                case 1:
                    Console.WriteLine("Hello!");
                    break;
                case 2:
                    Console.WriteLine(DateTime.Now.ToShortDateString());
                    break;
                case 3:
                    Console.WriteLine("Goodbye!");
                    break;
                default:
                    Console.WriteLine("Invalid choice.");
                    break;
            }
        } while (choice != 3);

        return 0;
    }
}

```

```
using System;

class Program
{
    static int Main(string[] args)
    {
        int[] marks = new int[5];
        for (int i = 0; i < marks.Length; i++)
        {
            Console.Write($"Enter mark for student {i + 1}: ");
            int.TryParse(Console.ReadLine(), out marks[i]);
        }
        Console.WriteLine("Student marks:");
        foreach (int m in marks)
        {
            Console.WriteLine(m);
        }
        int total = 0;
        int highest = marks[0];
        int lowest = marks[0];
        int passed = 0;
        int failed = 0;

        foreach (int m in marks)
        {
            total += m;
            if (m > highest) highest = m;
            if (m < lowest) lowest = m;

            if (m >= 50) passed++;
            else failed++;
        }
        double average = (double)total / marks.Length;
        Console.WriteLine($"Total: {total}");
        Console.WriteLine($"Average: {average}");
        Console.WriteLine($"Highest: {highest}");
        Console.WriteLine($"Lowest: {lowest}");
        Console.WriteLine($"Passed: {passed}");
        Console.WriteLine($"Failed: {failed}");

        return 0;
    }
}
```
